```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('gurgaon_properties_missing_value_imputation.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
latlong = pd.read_csv('latlong.csv')
```

```python
latlong
```

```python
latlong['latitude'] = latlong['coordinates'].str.split(',').str.get(0).str.split('°').str.get(0).astype('float')
```

```python
latlong['longitude'] = latlong['coordinates'].str.split(',').str.get(1).str.split('°').str.get(0).astype('float')
```

```python
latlong.head()
```

```python
new_df = df.merge(latlong, on='sector')
```

```python
new_df.columns
```

```python
group_df = new_df.groupby('sector')[['price','price_per_sqft','built_up_area','latitude','longitude']].mean()
```

```python
group_df
```

```python
fig = px.scatter_mapbox(group_df, lat="latitude", lon="longitude", color="price_per_sqft", size='built_up_area',
                  color_continuous_scale=px.colors.cyclical.IceFire, zoom=10,
                  mapbox_style="open-street-map",text=group_df.index)
fig.show()
```

```python
new_df.to_csv('data_viz1.csv',index=False)
```

```python
df1 = pd.read_csv('gurgaon_properties.csv')
```

```python
df1.head()
```

```python
wordcloud_df = df1.merge(df, left_index=True, right_index=True)[['features','sector']]
```

```python
wordcloud_df.head()
```

```python
sector_word=wordcloud_df.groupby(by='sector')
sector_word.get_group('sector 36').dropna()
```

```python
Sector_Word_df=pd.DataFrame(columns=['sector','features'])
Sector_Word_df

for name, group in sector_word:
    feature_list=[]
    for item in group['features'].dropna():
        feature_list.extend(item.replace("'",'').replace('[','').replace(']','').split(','))
    feature_word=' '.join(feature_list)
    new_row=pd.DataFrame({"sector":[name], "features":[feature_word]})
    Sector_Word_df=pd.concat([Sector_Word_df, new_row])
Sector_Word_df    
```

```python
import ast
main = []
for item in wordcloud_df['features'].dropna().apply(ast.literal_eval):
    main.extend(item)
```

```python
main
```

```python
from wordcloud import WordCloud
```

```python
feature_text = ' '.join(main)
```

```python
import pickle
pickle.dump(feature_text, open('feature_text.pkl','wb'))
```

```python
feature_text
```

```python
Sector_Word_df.to_csv('Sector_Word_df')
```

```python
plt.rcParams["font.family"] = "Arial"

wordcloud = WordCloud(width = 800, height = 800, 
                      background_color ='white', 
                      stopwords = set(['s']),  # Any stopwords you'd like to exclude
                      min_font_size = 10).generate(feature_text)

plt.figure(figsize = (8, 8), facecolor = None) 
plt.imshow(wordcloud, interpolation='bilinear') 
plt.axis("off") 
plt.tight_layout(pad = 0) 
plt.show() # st.pyplot()
```
